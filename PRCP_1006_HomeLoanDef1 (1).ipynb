```python
# Load All Tables
bureau = pd.read_csv("bureau.csv")
bureau_balance = pd.read_csv("bureau_balance.csv")
pos = pd.read_csv("POS_CASH_balance.csv")
cc = pd.read_csv("credit_card_balance.csv")
prev = pd.read_csv("previous_application.csv")
install = pd.read_csv("installments_payments.csv")
```

```python
# Bureau Balance → Bureau
bb_agg = bureau_balance.groupby('SK_ID_BUREAU').agg({
    'MONTHS_BALANCE': ['min', 'max'],
    'STATUS': lambda x: (x != '0').sum()
})
bb_agg.columns = ['BB_MONTH_MIN','BB_MONTH_MAX','BB_LATE_COUNT']
bb_agg.reset_index(inplace=True)

bureau = bureau.merge(bb_agg, on='SK_ID_BUREAU', how='left')
```

```python
# Bureau → Application
bureau_agg = bureau.groupby('SK_ID_CURR').agg({
    'CREDIT_ACTIVE': 'count',
    'AMT_CREDIT_SUM': 'sum',
    'AMT_CREDIT_SUM_DEBT': 'sum',
    'AMT_CREDIT_SUM_OVERDUE': 'sum',
    'BB_LATE_COUNT': 'sum'
})

bureau_agg.columns = [
    'BUREAU_LOAN_COUNT',
    'BUREAU_TOTAL_CREDIT',
    'BUREAU_TOTAL_DEBT',
    'BUREAU_TOTAL_OVERDUE',
    'BUREAU_LATE_PAYMENTS'
]

bureau_agg.reset_index(inplace=True)
```

```python
# POS_CASH_balance Aggregation
pos_agg = pos.groupby('SK_ID_CURR').agg({
    'MONTHS_BALANCE': ['min','max'],
    'SK_DPD': 'mean',
    'SK_DPD_DEF': 'mean'
})

pos_agg.columns = ['POS_MIN_MONTH','POS_MAX_MONTH','POS_DPD','POS_DPD_DEF']
pos_agg.reset_index(inplace=True)
```

```python
# Credit Card Balance Aggregation
cc_agg = cc.groupby('SK_ID_CURR').agg({
    'AMT_BALANCE': 'mean',
    'AMT_CREDIT_LIMIT_ACTUAL': 'mean',
    'SK_DPD': 'mean'
})

cc_agg.columns = ['CC_AVG_BAL','CC_CREDIT_LIMIT','CC_DPD']
cc_agg.reset_index(inplace=True)
```

```python
# Previous Application Aggregation
prev_agg = prev.groupby('SK_ID_CURR').agg({
    'AMT_APPLICATION': 'mean',
    'AMT_CREDIT': 'mean',
    'NAME_CONTRACT_STATUS': lambda x: (x == 'Approved').sum()
})

prev_agg.columns = ['PREV_APP_AMT','PREV_CREDIT_AMT','PREV_APPROVED_COUNT']
prev_agg.reset_index(inplace=True)
```

```python
# Installments Payments Aggregation
install['PAYMENT_DIFF'] = install['AMT_PAYMENT'] - install['AMT_INSTALMENT']
install['LATE_PAYMENT'] = install['DAYS_ENTRY_PAYMENT'] - install['DAYS_INSTALMENT']

install_agg = install.groupby('SK_ID_CURR').agg({
    'PAYMENT_DIFF': 'mean',
    'LATE_PAYMENT': 'mean'
})

install_agg.columns = ['AVG_PAYMENT_DIFF','AVG_LATE_DAYS']
install_agg.reset_index(inplace=True)
```

```python
# Final Merge
final_df = app.merge(bureau_agg, on='SK_ID_CURR', how='left')
final_df = final_df.merge(pos_agg, on='SK_ID_CURR', how='left')
final_df = final_df.merge(cc_agg, on='SK_ID_CURR', how='left')
final_df = final_df.merge(prev_agg, on='SK_ID_CURR', how='left')
final_df = final_df.merge(install_agg, on='SK_ID_CURR', how='left')

print("Final Shape:", final_df.shape)
final_df.head()
```

```python
# Data Preparation
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix, classification_report
)
import warnings
warnings.filterwarnings("ignore")

df = final_df.copy()

# Separate target
X = df.drop(['TARGET'], axis=1)
y = df['TARGET']

# Encode categorical variables
for col in X.select_dtypes(include='object').columns:
    X[col] = LabelEncoder().fit_transform(X[col].astype(str))

# Fill missing values
X = X.fillna(X.median())

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
# Evaluation Function (Reusable)
def evaluate_model(model, X_test, y_test):
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:,1]

    return {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred),
        "ROC_AUC": roc_auc_score(y_test, y_prob)
    }
```

```python
# Logistic Regression (Baseline)
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(max_iter=1000, class_weight='balanced')
log_reg.fit(X_train, y_train)

log_reg_results = evaluate_model(log_reg, X_test, y_test)
log_reg_results
```

```python
# Random Forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    random_state=42,
    class_weight='balanced'
)

rf.fit(X_train, y_train)
rf_results = evaluate_model(rf, X_test, y_test)
rf_results
```

```python
# Gradient Boosting
from sklearn.ensemble import GradientBoostingClassifier

gb = GradientBoostingClassifier(
    n_estimators=150,
    learning_rate=0.05,
    random_state=42
)

gb.fit(X_train, y_train)
gb_results = evaluate_model(gb, X_test, y_test)
gb_results
```

```python
# XGBoost
from xgboost import XGBClassifier

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight= (y_train.value_counts()[0] / y_train.value_counts()[1]),
    eval_metric='logloss',
    random_state=42
)

xgb.fit(X_train, y_train)
xgb_results = evaluate_model(xgb, X_test, y_test)
xgb_results
```

```python
# Model Comparison Table
model_comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest", "Gradient Boosting", "XGBoost"],
    "Accuracy": [
        log_reg_results["Accuracy"],
        rf_results["Accuracy"],
        gb_results["Accuracy"],
        xgb_results["Accuracy"]
    ],
    "Precision": [
        log_reg_results["Precision"],
        rf_results["Precision"],
        gb_results["Precision"],
        xgb_results["Precision"]
    ],
    "Recall": [
        log_reg_results["Recall"],
        rf_results["Recall"],
        gb_results["Recall"],
        xgb_results["Recall"]
    ],
    "ROC_AUC": [
        log_reg_results["ROC_AUC"],
        rf_results["ROC_AUC"],
        gb_results["ROC_AUC"],
        xgb_results["ROC_AUC"]
    ]
})

model_comparison.sort_values(by="ROC_AUC", ascending=False)
```

```python
# Confusion Matrix (Gradient Boosting)
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Predictions
gb_pred = gb.predict(X_test)

# Confusion Matrix
cm = confusion_matrix(y_test, gb_pred)

plt.figure(figsize=(4,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Confusion Matrix – Gradient Boosting")
plt.show()
```

# Report on Challenges Faced & Techniques Used

## 1. Missing Values in Dataset

    Challenge:

* The dataset contained a large number of missing values, especially in:

* External bureau-related features

* Credit card and POS balance data

* Some features had more than 40–60% missing values, which could bias the model or reduce usable data if rows were dropped.

    Technique Used:

* Median imputation for numerical variables

* Mode imputation / label encoding handling for categorical variables

   Reason:

* Median is robust to outliers and skewed financial distributions.
* Dropping rows would lead to significant data loss.


## 2. Highly Imbalanced Target Variable

   Challenge:

* The target variable (TARGET) was imbalanced, with significantly fewer defaulters compared to non-defaulters.

* This could lead to:

* High accuracy but poor detection of defaulters

* Business risk due to false approvals

  Technique Used:

* Used class_weight = 'balanced' for Logistic Regression and Random Forest

* Used scale_pos_weight in XGBoost

* Evaluated models using Recall and ROC-AUC, not just accuracy

    Reason:
* In banking, identifying defaulters (Recall) is more important than overall accuracy.


## 3. High Dimensionality & Multiple Data Sources

   Challenge:

* The data came from multiple relational tables:

* bureau

* bureau_balance

* POS_CASH_balance

* credit_card_balance

* previous_application

* installments_payments

* Each table had one-to-many relationships, increasing complexity.

    Technique Used:

* Performed aggregation at customer level (SK_ID_CURR)

* Created meaningful statistical features (mean, sum, count)

* Merged all aggregated tables into a single modeling dataset

  Reason:

* Machine learning models require flat, customer-level data.
* Aggregation preserves historical behavior without exploding rows.


## 4. Non-Linear Relationships in Financial Data

     Challenge:

* Customer default behavior depends on complex, non-linear interactions:

* Income vs loan amount

* Credit usage patterns

* Repayment delays

* Linear models alone could not capture these patterns effectively.

     Technique Used:

* Used tree-based ensemble models:

* Random Forest

* Gradient Boosting

* XGBoost

   Reason:
* Tree-based models handle non-linearity, interactions, and missing values efficiently.


## 5.Feature Scaling & Skewed Distributions
    Challenge:

* Financial features such as income, credit amount, and debt were highly skewed.

    Technique Used:

* Visualized distributions using log-scale plots

* Used models (tree-based) that are not sensitive to feature scaling

  Reason:
* Avoiding unnecessary scaling simplifies pipeline and preserves interpretability.


## 6. Model Selection for Business Use
   Challenge:

* Selecting a model that balances:

* Performance

* Business risk

* Scalability

* Interpretability

* Technique Used

Built and compared four models:

* Logistic Regression

* Random Forest

* Gradient Boosting

* XGBoost

* Compared using ROC-AUC, Recall, Precision

  Reason:
* ROC-AUC measures ranking ability, while Recall minimizes risky approvals.


## 7. Computational Cost & Training Time

   Challenge:

* Large dataset size increased training time, especially for ensemble models.

   Technique Used

* Controlled tree depth

* Tuned number of estimators

* Used efficient boosting methods (XGBoost)

  Reason:
* Optimized performance without excessive computational cost.


# Final Summary

The major challenges in this project included missing values, class imbalance, high dimensionality, and non-linear relationships in financial data. These challenges were addressed using robust preprocessing, feature aggregation, class balancing techniques, and advanced ensemble models. XGBoost emerged as the most suitable production model due to its superior predictive performance and ability to handle real-world banking data complexities.
